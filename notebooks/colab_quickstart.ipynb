```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('results.csv')
plt.plot(df.new_tokens, df.cache_tps, marker='o', label='with KV cache')
plt.plot(df.new_tokens, df.nocache_tps, marker='o', label='no cache')
plt.xlabel('new tokens generated'); plt.ylabel('tokens/sec'); plt.legend(); plt.grid(alpha=.3)
plt.savefig('speedup.png', dpi=150, bbox_inches='tight'); plt.show()
```
