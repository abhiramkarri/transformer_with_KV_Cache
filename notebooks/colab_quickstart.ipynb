```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('results.csv')
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
for col, label in [('nocache_tps','no cache'), ('cache_tps','cat cache'), ('static_tps','static cache')]:
    ax[0].plot(df.new_tokens, df[col], marker='o', label=label)
ax[0].set_xlabel('new tokens generated'); ax[0].set_ylabel('tokens/sec'); ax[0].legend(); ax[0].grid(alpha=.3)
for col, label in [('nocache_mb','no cache'), ('cat_mb','cat cache'), ('static_mb','static cache')]:
    ax[1].plot(df.new_tokens, df[col], marker='o', label=label)
ax[1].set_xlabel('new tokens generated'); ax[1].set_ylabel('extra peak GPU memory (MB)'); ax[1].legend(); ax[1].grid(alpha=.3)
plt.savefig('speedup.png', dpi=150, bbox_inches='tight'); plt.show()
```
